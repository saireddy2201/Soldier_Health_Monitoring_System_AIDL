```python

# Select input features
features = [
    "Temperature",
    "Heart Rate",
    "Pulse",
    "BPSYS",
    "BPDIA",
    "Respiratory Rate",
    "Oxygen Saturation",
    "PH"
]

# Input and output
X = df[features].copy()
y = df["Causes Respiratory Imbalance"].copy()

print("Input Features:")
print(X.head())

print("\nTarget Categories:")
print(y.unique())

print("\nInput Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

# Encode target categories
encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y)

print("Class Mapping:")
for i, label in enumerate(encoder.classes_):
    print(i, "=", label)

# Split training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.2,
    random_state=42,
    stratify=y_encoded
)

# Normalize input features
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print("\nTraining Records:", X_train.shape[0])
print("Testing Records:", X_test.shape[0])
print("Preprocessing completed!")
```

```python

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input

# Create ANN model
model = Sequential([
    Input(shape=(8,)),

    Dense(64, activation="relu"),
    Dropout(0.2),

    Dense(32, activation="relu"),

    Dense(16, activation="relu"),

    Dense(4, activation="softmax")
])

# Compile model
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Display model
model.summary()
```

```python

history = model.fit(
    X_train,
    y_train,
    epochs=30,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)

print("Model Training Completed!")
```

```python

plt.figure(figsize=(10, 5))

plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")

plt.title("ANN Model Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.legend()
plt.grid()
plt.show()
```

```python

plt.figure(figsize=(10, 5))

plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")

plt.title("ANN Model Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.grid()
plt.show()
```

```python

from sklearn.metrics import classification_report

y_pred = model.predict(X_test)
y_pred_classes = np.argmax(y_pred, axis=1)

print(classification_report(
    y_test,
    y_pred_classes,
    labels=np.arange(len(encoder.classes_)),
    target_names=encoder.classes_,
    zero_division=0
))
```

```python

from sklearn.metrics import confusion_matrix

cm = confusion_matrix(
    y_test,
    y_pred_classes,
    labels=np.arange(len(encoder.classes_))
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=encoder.classes_,
    yticklabels=encoder.classes_
)

plt.title("Soldier Health Monitoring - Confusion Matrix")
plt.xlabel("Predicted Category")
plt.ylabel("Actual Category")
plt.show()
```

```python

from sklearn.utils.class_weight import compute_class_weight
import numpy as np

classes = np.unique(y_train)

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y_train
)

class_weights = dict(zip(classes, weights))

print("Class Weights:")
for key, value in class_weights.items():
    print(encoder.classes_[key], ":", value)
```

```python

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.callbacks import EarlyStopping

model = Sequential([
    Input(shape=(8,)),
    Dense(64, activation="relu"),
    Dropout(0.3),
    Dense(32, activation="relu"),
    Dropout(0.2),
    Dense(16, activation="relu"),
    Dense(4, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

early_stop = EarlyStopping(
    monitor="val_loss",
    patience=6,
    restore_best_weights=True
)

history = model.fit(
    X_train,
    y_train,
    epochs=50,
    batch_size=32,
    validation_split=0.2,
    class_weight=class_weights,
    callbacks=[early_stop],
    verbose=1
)

print("Improved model training completed!")
```

```python

loss, accuracy = model.evaluate(X_test, y_test)

print("New Test Accuracy:", accuracy * 100, "%")

y_pred = model.predict(X_test)
y_pred_classes = np.argmax(y_pred, axis=1)

from sklearn.metrics import classification_report

print(classification_report(
    y_test,
    y_pred_classes,
    target_names=encoder.classes_,
    zero_division=0
))
```

```python

import gradio as gr
import numpy as np

def predict_health(temp, heart, pulse, sys_bp,
                   dia_bp, respiratory, oxygen, ph):

    readings = np.array([[
        temp, heart, pulse, sys_bp,
        dia_bp, respiratory, oxygen, ph
    ]], dtype=float)

    scaled_data = scaler.transform(readings)

    probabilities = model.predict(
        scaled_data, verbose=0
    )[0]

    predicted_class = np.argmax(probabilities)

    result = encoder.inverse_transform(
        [predicted_class]
    )[0]

    confidence = probabilities[predicted_class] * 100

    return (
        f"Predicted Category: {result}",
        f"Model Confidence: {confidence:.2f}%"
    )


demo = gr.Interface(
    fn=predict_health,

    inputs=[
        gr.Number(label="Temperature", value=98),
        gr.Number(label="Heart Rate", value=90),
        gr.Number(label="Pulse", value=75),
        gr.Number(label="Systolic BP", value=120),
        gr.Number(label="Diastolic BP", value=80),
        gr.Number(label="Respiratory Rate", value=18),
        gr.Number(label="Oxygen Saturation", value=0.98),
        gr.Number(label="PH", value=7)
    ],

    outputs=[
        gr.Textbox(label="Health Prediction"),
        gr.Textbox(label="Prediction Confidence")
    ],

    title="SOLDIER HEALTH MONITORING SYSTEM",
    description="AI-Based Soldier Health Monitoring and Prediction System"
)

demo.launch(
    share=True,
    debug=True,
    theme=gr.themes.Soft()
)
```

```python

import gradio as gr
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import tempfile
from datetime import datetime

# -----------------------------------------
# 1. PROJECT CONFIGURATION
# -----------------------------------------

features = [
    "Temperature",
    "Heart Rate",
    "Pulse",
    "BPSYS",
    "BPDIA",
    "Respiratory Rate",
    "Oxygen Saturation",
    "PH"
]

target = "Causes Respiratory Imbalance"

# Check required variables and columns
required_columns = features + [target]
missing = [c for c in required_columns if c not in df.columns]

if missing:
    raise ValueError(f"Missing dataset columns: {missing}")

if model is None or scaler is None or encoder is None:
    raise ValueError("Train the ANN model before launching the dashboard.")

# Ranges observed in the training dataset
data_min = df[features].min()
data_max = df[features].max()

# -----------------------------------------
# 2. ALERT MESSAGES
# -----------------------------------------

alert_messages = {
    "Normal":
        "🟢 **MODEL RESULT: NORMAL**\n\n"
        "The model predicts the Normal category. "
        "This does not guarantee that the person is healthy.",

    "Mild":
        "🟡 **MODEL RESULT: MILD**\n\n"
        "The model predicts the Mild category. "
        "Review the readings and the person's condition.",

    "Chronic":
        "🟠 **MODEL RESULT: CHRONIC**\n\n"
        "The model predicts the Chronic category. "
        "Further assessment by qualified health personnel is advisable.",

    "Severe":
        "🔴 **MODEL RESULT: SEVERE**\n\n"
        "The model predicts the Severe category. "
        "This prototype is not an emergency service. "
        "Follow established medical or emergency procedures when needed."
}

# -----------------------------------------
# 3. CREATE DATASET ANALYTICS GRAPH
# -----------------------------------------

fig_dataset, ax = plt.subplots(figsize=(7, 4))

category_counts = df[target].value_counts()

ax.bar(
    category_counts.index.astype(str),
    category_counts.values
)

ax.set_title("Dataset Health Category Distribution")
ax.set_xlabel("Health Category")
ax.set_ylabel("Number of Records")
ax.grid(axis="y", alpha=0.25)

plt.tight_layout()

# -----------------------------------------
# 4. CREATE PREDICTION GRAPH
# -----------------------------------------

def create_probability_graph(probabilities):

    fig, ax = plt.subplots(figsize=(7, 4))

    classes = encoder.classes_

    ax.bar(classes, probabilities * 100)

    ax.set_title("AI Prediction Probability")
    ax.set_xlabel("Health Category")
    ax.set_ylabel("Estimated Probability (%)")

    ax.set_ylim(0, 100)
    ax.grid(axis="y", alpha=0.25)

    plt.tight_layout()

    return fig

# -----------------------------------------
# 5. PREDICTION FUNCTION
# -----------------------------------------

def predict_soldier(
    temperature,
    heart_rate,
    pulse,
    systolic,
    diastolic,
    respiratory,
    oxygen,
    ph,
    previous_history
):

    values = [
        temperature,
        heart_rate,
        pulse,
        systolic,
        diastolic,
        respiratory,
        oxygen,
        ph
    ]

    if any(v is None or not np.isfinite(float(v)) for v in values):
        raise gr.Error("Please enter valid numbers in all eight fields.")

    # Prepare input using the exact training feature order
    input_df = pd.DataFrame([values], columns=features)

    # Check whether readings are within the dataset's observed ranges
    outside = []

    for col, val in zip(features, values):

        if val < data_min[col] or val > data_max[col]:

            outside.append(
                f"{col}: entered {val}, "
                f"observed range {data_min[col]} to {data_max[col]}"
            )

    # Scale input
    scaled_input = scaler.transform(input_df)

    # ANN prediction
    probabilities = model.predict(
        scaled_input,
        verbose=0
    )[0]

    predicted_index = int(np.argmax(probabilities))

    predicted_category = str(
        encoder.inverse_transform([predicted_index])[0]
    )

    confidence = float(probabilities[predicted_index] * 100)

    # Prediction output
    prediction_text = (
        f"## Predicted Category: {predicted_category}"
    )

    confidence_text = (
        f"### Estimated Class Probability: {confidence:.2f}%"
    )

    # Generate warning
    alert_text = alert_messages.get(
        predicted_category,
        "Model returned an unknown category."
    )

    if outside:

        alert_text += (
            "\n\n⚠️ **INPUT RANGE WARNING**\n\n"
            "One or more readings are outside the ranges observed "
            "in the training dataset. This prediction may be unreliable.\n\n"
            + "\n".join(outside)
        )

    # -----------------------------------------
    # 6. CREATE PREDICTION HISTORY
    # -----------------------------------------

    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    record = {
        "Timestamp": timestamp
    }

    for col, val in zip(features, values):
        record[col] = val

    record["Predicted Category"] = predicted_category
    record["Estimated Probability (%)"] = round(confidence, 2)

    # Include probability for every category
    for category, probability in zip(
        encoder.classes_,
        probabilities
    ):
        record[f"{category} Probability (%)"] = round(
            float(probability) * 100, 2
        )

    new_history = (previous_history or []) + [record]

    history_df = pd.DataFrame(new_history)

    history_view = history_df[
        [
            "Timestamp",
            "Predicted Category",
            "Estimated Probability (%)"
        ]
    ]

    # -----------------------------------------
    # 7. CREATE ANALYTICS SUMMARY
    # -----------------------------------------

    total = len(new_history)

    counts = history_df["Predicted Category"].value_counts()

    analytics = (
        "## Session Analytics\n\n"
        f"**Total predictions:** {total}\n\n"
    )

    for category in encoder.classes_:

        analytics += (
            f"- **{category}:** {counts.get(category, 0)} predictions\n"
        )

    # -----------------------------------------
    # 8. CREATE DOWNLOADABLE CSV REPORT
    # -----------------------------------------

    report_df = pd.DataFrame([record])

    report_df["Important Note"] = (
        "Educational prototype only; not a medical diagnosis."
    )

    temp_file = tempfile.NamedTemporaryFile(
        suffix=".csv",
        prefix="soldier_health_report_",
        delete=False
    )

    temp_file.close()

    report_df.to_csv(temp_file.name, index=False)

    # -----------------------------------------
    # 9. CREATE PROBABILITY GRAPH
    # -----------------------------------------

    probability_fig = create_probability_graph(probabilities)

    return (
        prediction_text,
        alert_text,
        confidence_text,
        probability_fig,
        history_view,
        analytics,
        temp_file.name,
        new_history
    )

# -----------------------------------------
# 10. DOWNLOAD FULL SESSION HISTORY
# -----------------------------------------

def download_session(history):

    if not history:
        raise gr.Error("No prediction history available yet.")

    history_df = pd.DataFrame(history)

    temp_file = tempfile.NamedTemporaryFile(
        suffix=".csv",
        prefix="soldier_session_history_",
        delete=False
    )

    temp_file.close()

    history_df.to_csv(temp_file.name, index=False)

    return temp_file.name

# -----------------------------------------
# 11. CLEAR SESSION HISTORY
# -----------------------------------------

def clear_history():

    empty_df = pd.DataFrame(
        columns=[
            "Timestamp",
            "Predicted Category",
            "Estimated Probability (%)"
        ]
    )

    return (
        [],
        empty_df,
        "## Session Analytics\n\nNo predictions recorded yet.",
        None,
        None,
        "",
        "",
        ""
    )

# -----------------------------------------
# 12. BUILD UPGRADED GRADIO DASHBOARD
# -----------------------------------------

custom_css = """
.gradio-container {
    max-width: 1200px !important;
}

#main-title {
    text-align: center;
    font-size: 30px;
    font-weight: bold;
}

#project-subtitle {
    text-align: center;
    opacity: 0.8;
}

"""

empty_history = pd.DataFrame(
    columns=[
        "Timestamp",
        "Predicted Category",
        "Estimated Probability (%)"
    ]
)

with gr.Blocks(title="Soldier Health Monitoring System") as demo:

    history_state = gr.State([])

    gr.Markdown(
        "# 🪖 SOLDIER HEALTH MONITORING SYSTEM",
        elem_id="main-title"
    )

    gr.Markdown(
        "AI-Based Soldier Health Prediction, Alerts and Analytics",
        elem_id="project-subtitle"
    )

    gr.Markdown(
        "Enter readings using the same units and scales as the training dataset."
    )

    with gr.Row():

        # ---------------------------------
        # LEFT SIDE: INPUTS
        # ---------------------------------

        with gr.Column():

            gr.Markdown("## Soldier Health Parameters")

            temperature = gr.Number(
                label="Temperature (dataset scale)",
                value=float(df["Temperature"].median())
            )

            heart_rate = gr.Number(
                label="Heart Rate",
                value=float(df["Heart Rate"].median())
            )

            pulse = gr.Number(
                label="Pulse",
                value=float(df["Pulse"].median())
            )

            systolic = gr.Number(
                label="Systolic BP",
                value=float(df["BPSYS"].median())
            )

            diastolic = gr.Number(
                label="Diastolic BP",
                value=float(df["BPDIA"].median())
            )

            respiratory = gr.Number(
                label="Respiratory Rate",
                value=float(df["Respiratory Rate"].median())
            )

            oxygen = gr.Number(
                label="Oxygen Saturation (dataset scale)",
                value=float(df["Oxygen Saturation"].median()),
                precision=5
            )

            ph = gr.Number(
                label="PH (dataset scale)",
                value=float(df["PH"].median())
            )

            predict_btn = gr.Button(
                "🔍 Analyze Soldier Health",
                variant="primary"
            )

            clear_btn = gr.Button(
                "Clear Session History"
            )

        # ---------------------------------
        # RIGHT SIDE: RESULTS AND ALERTS
        # ---------------------------------

        with gr.Column():

            gr.Markdown("## AI Prediction Results")

            prediction_output = gr.Markdown()

            alert_output = gr.Markdown()

            confidence_output = gr.Markdown()

            probability_plot = gr.Plot(
                label="Prediction Probability"
            )

            report_file = gr.File(
                label="Download Current Prediction Report",
                interactive=False
            )

    # ---------------------------------
    # ANALYTICS SECTION
    # ---------------------------------

    gr.Markdown("---")
    gr.Markdown("# 📊 Analytics Dashboard")

    with gr.Row():

        analytics_output = gr.Markdown(
            "## Session Analytics\n\nNo predictions recorded yet."
        )

        dataset_plot = gr.Plot(
            value=fig_dataset,
            label="Dataset Category Distribution"
        )

    # ---------------------------------
    # HISTORY SECTION
    # ---------------------------------

    gr.Markdown("---")
    gr.Markdown("# 📋 Prediction History")

    history_table = gr.Dataframe(
        value=empty_history,
        interactive=False,
        wrap=True
    )

    download_history_btn = gr.Button(
        "⬇ Download Complete Session History",
        variant="primary"
    )

    session_report_file = gr.File(
        label="Download Session History CSV",
        interactive=False
    )

    gr.Markdown(
        """
        **Important:** This system is an educational AI prototype.
        Its predictions and estimated probabilities are not medical diagnoses
        or validated emergency alerts. Sensor units and model performance
        must be validated before any real-world use.
        """
    )

    # ---------------------------------
    # BUTTON ACTIONS
    # ---------------------------------

    predict_btn.click(
        fn=predict_soldier,

        inputs=[
            temperature,
            heart_rate,
            pulse,
            systolic,
            diastolic,
            respiratory,
            oxygen,
            ph,
            history_state
        ],

        outputs=[
            prediction_output,
            alert_output,
            confidence_output,
            probability_plot,
            history_table,
            analytics_output,
            report_file,
            history_state
        ]
    )

    download_history_btn.click(
        fn=download_session,
        inputs=history_state,
        outputs=session_report_file
    )

    clear_btn.click(
        fn=clear_history,
        inputs=[],
        outputs=[
            history_state,
            history_table,
            analytics_output,
            probability_plot,
            report_file,
            session_report_file,
            prediction_output,
            alert_output
        ]
    )

# -----------------------------------------
# 13. LAUNCH APPLICATION
# -----------------------------------------

demo.launch(
    share=True,
    debug=True,
    theme=gr.themes.Soft(),
    css=custom_css
)
```
